```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ","_")
```

```python
# Step 4A: Convert units (multiply 1000_ha and 1000_tons columns by 1000)
for col in df.columns:
    if "area_(1000_ha)" in col or "production_(1000_tons)" in col:
        df[col] = df[col] * 1000

# Step 4B: Rename columns to base units (ha and tons)
df.rename(columns=lambda x: x.replace("_(1000_ha)", "_(ha)").replace("_(1000_tons)", "_(tons)"), inplace=True)

# Confirm that the renaming worked
print([col for col in df.columns if "1000" in col])  # should return []
```

```python
# Select all numeric columns
numeric_cols= df.select_dtypes(include=['float64','int64']).columns
```

```python
# Check how many negative values are in each numeric column
negatives = (df[numeric_cols] < 0).sum()
print(negatives[negatives > 0])
```

```python
import numpy as np
# Replace negative values with NaN
df[numeric_cols] = df[numeric_cols].mask(df[numeric_cols] < 0,np.nan)
```

```python
df.head()
```

```python
 #Step 5B: Review extremely high yield values (>10,000 kg/ha)
yield_cols =[col for col in df.columns if 'yield_(kg_per_ha)' in col]

for col in yield_cols:
    high_yield_rows = df[df[col] > 10000][['state_name','dist_name','year',col]]
    if not high_yield_rows.empty:
        print(f"\n⚠️ High values found in '{col}':")
        print(high_yield_rows)
```

```python
for col in yield_cols:
    # Identitfy rows with yield > 10000
    high_yield_mask = df[col] > 10000

    if high_yield_mask.any():
        print(f"🔧 Fixing high values in: {col}")
    
```

```python
# Replace each high value with the mean yield for that crop in the same state 
for idx in high_yield_mask.index:
    state = df.at[idx,'state_name']
     # Mean yield in that state, ignoring high outliers and NaNs
    state_mean = df[(df['state_name'] == state) & (df[col] <= 10000)][col].mean()
    df.at[idx, col] = state_mean
```

```python
#1. Check for Any Remaining Negative Values
print("Negative values remaining:",(df[numeric_cols] < 0).sum().sum())
```

```python
 #2.Check for Any Yield Values Still > 10,000
for col in yield_cols:
    max_val = df[col].max()
    if max_val > 10000:
        print(f" Still high in {col}: {max_val}")
```

```python
for col in yield_cols:
    high_yield_mask = df[col] > 10000
    if high_yield_mask.any():
        print(f"🔧 Replacing high values in: {col}")
        
        # Loop through each row with a high yield
        for idx in df[high_yield_mask].index:
            state = df.at[idx, 'state_name']
            
            # Calculate mean of valid (≤10,000) values in same state
            valid_state_yield = df[(df['state_name'] == state) & (df[col] <= 10000)][col]
            state_mean = valid_state_yield.mean()

            # Only replace if state_mean is not NaN
            if pd.notnull(state_mean):
                df.at[idx, col] = state_mean
            else:
                # If state mean is not available, replace with global mean below threshold
                global_mean = df[df[col] <= 10000][col].mean()
                df.at[idx, col] = global_mean
```

```python
# Check for Unrealistic Outliers in Yields (>10,000 kg/ha)
yield_cols = [col for col in df.columns if "yield" in col.lower()]
for col in yield_cols:
    max_val = df[col].max()
    if max_val > 10000:
        print(f"⚠️ High value in {col}: {max_val}")
```

```python
for col in yield_cols:
    max_val = df[col].max()
    print(f"✅ Max in {col}: {max_val}")
```

```python
print("🧪 Null values after cleaning:")
print(df[numeric_cols].isnull().sum().sort_values(ascending=False).head(10))
```

```python
for col in df.columns:
    if df[col].isnull().sum() > 0:
        df[col] = df.groupby('state_name')[col].transform(lambda x: x.fillna(x.mean()))
```
